**IMPORTING NEEDED LIBRARIES**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

**LOAD DATASET**

```python
data=pd.read_csv("Insurance.csv")
```

**DATA PREPROCESSING**

```python
data.head()
```

```python
data.isnull().sum()
```

```python
sns.boxplot(data=data)
plt.title('Before removal of outliers')
plt.show()
```

**Why we didn't remove outliers?**

Box plots were used to identify potential outliers in the dataset. However, these outliers were not removed because they represent genuine real-world cases, especially in insurance data where extreme values naturally occur. Retaining these values helps preserve data realism and improves model robustness.

**MODEL CREATION STEPS**

```python
x=data[['age','sex','bmi','children','smoker','region','charges']]
y=data['insuranceclaim']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=RandomForestClassifier(n_estimators=1000,random_state=42)
model.fit(x_train,y_train)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model2=LogisticRegression(max_iter=1000)
model2.fit(x_train,y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model3=DecisionTreeClassifier(max_depth=500)
model3.fit(x_train,y_train)
```

**CHECK FOR MODEL ACCURACY**

```python
y_pred=model.predict(x_test)
y_pred
```

```python
y2_pred=model2.predict(x_test)      #for logisticRegression
```

```python
y3_pred=model3.predict(x_test)      #for DecisionTree
```

```python
accuracy=accuracy_score(y_pred,y_test)
print("Random forest accuracy:",accuracy)
accuracy2=accuracy_score(y2_pred,y_test)
print("Logistic Regression accuracy:",accuracy2)
accuracy3=accuracy_score(y3_pred,y_test)
print("Decision Tree accuracy:",accuracy3)
```

```python
y=[accuracy,accuracy2,accuracy3]
x=['RandomForestClassifier','logisticRegression','DecisionTree']
plt.plot(x,y,marker='o' ,color='purple',linestyle='--')
plt.title('ACCURACY LINE PLOT FOR VARIOUS MODEL')
plt.xlabel('Classification algorithm')
plt.ylabel("Accuracy")
plt.show()
```

Decision Tree shows higher training accuracy but may **overfit and perform poorly on new data**. **Random Forest,** though slightly lower in accuracy, is more stable, reliable, and explainable for **real-world predictions**.

**GOOD FOR PREDICTION**

An accuracy of 91% indicates that the model correctly predicts the insurance claim outcome for the majority of cases, demonstrating strong predictive performance. This level of accuracy suggests that the selected features and learning algorithm effectively capture the underlying patterns in the data, making the model reliable for practical use.

```python
print(classification_report(y_test,y_pred,target_names=['not claim','claim']))
```

```python
conf_matrix=confusion_matrix(y_test,y_pred)
sns.heatmap(conf_matrix,annot=True,fmt='d',cmap='Greens',
            xticklabels=['NOT CLAIM','CLAIM'],
            yticklabels=['NOT CLAIM','CLAIM'])
plt.xlabel('predicted')
plt.ylabel('actual')
plt.title('Confusion matrix')
plt.show()
```
